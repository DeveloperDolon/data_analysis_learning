```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
%matplotlib inline
```

```python
pokemon = pd.read_csv("pokemon.csv")
print(pokemon.shape)
pokemon.head(10)
```

```python
plt.hist(data=pokemon, x="speed")
plt.show()
```

```python
plt.hist(data=pokemon, x="speed", bins=20)
plt.show()
```

```python
bins = np.arange(0, pokemon["speed"].max() + 5, 5)
plt.hist(data=pokemon, x="speed", bins=bins)
plt.show()
```

```python
sb.histplot(data=pokemon, x="speed", kde=True, stat="density", bins=20)
plt.show()

sb.histplot(data=pokemon, x="speed")
plt.show()
```

```python
bin_edges = np.arange(0, pokemon["speed"].max() + 1, 5)
sb.histplot(data=pokemon, x="speed", bins=bin_edges, alpha=1)
plt.show()
```

```python
plt.figure(figsize=[20, 5])
plt.subplot(1, 2, 1)
bins = np.arange(0, pokemon["speed"].max() + 4, 4)
plt.hist(data=pokemon, x="speed", bins=bins)

plt.subplot(1, 2, 2)
bins = np.arange(0, pokemon["speed"].max() + 0.25, 0.25)
plt.hist(data=pokemon, x="speed", bins=bins)
plt.show()

plt.subplot(1, 2, 1)
bins = np.arange(0, pokemon["speed"].max() + 5, 2)
plt.hist(data=pokemon, x="speed", bins=bins)
plt.show()
```
