```python
import pandas as pd
```

```python
items = [
    {"bikes": 20, "pants": 35, "watches": 35}, 
    {"watches": 10, "glasses": 50, "bikes": 15, "pants": 5},
]

store_items = pd.DataFrame(items, index=["Store 1", "Store 2"])
store_items
```

```python
print()
print('How many bikes are in each store:\n', store_items[['bikes']])
```

```python
print()
print('How many bikes and pants are in each store:\n', store_items[['bikes', 'pants']])
print()
```

```python
print()
print('How many bikes are in Store 2:', store_items['bikes']['Store 2'])
```

```python
store_items['shirts'] = [15, 2]

# We display the modified DataFrame
print(store_items)
```

```python
store_items['suits'] = store_items['pants'] + store_items['shirts']

# We display the modified DataFrame
print(store_items)
```

```python
# We create a dictionary from a list of Python dictionaries that will contain the number of different items at the new store
new_items = [{'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4}]

# We create new DataFrame with the new_items and provide and index labeled store 3
new_store = pd.DataFrame(new_items, index = ['store 3'])

# We display the items at the new store
print(new_store)
```

```python
# We append store 3 to our store_items DataFrame
store_items = pd.concat([store_items, new_store])

# We display the modified DataFrame
print(store_items)
```

```python
# We add a new column using data from particular rows in the watches column
store_items['new watches'] = store_items['watches'][1:]

# We display the modified DataFrame
print(store_items)
```

```python
# We insert a new column with label shoes right before the column with numerical index 5
store_items.insert(5, 'shoes', [8, 5, 0])

# we display the modified DataFrame
print(store_items)
```

```python
# We remove the new watches column
store_items.pop('new watches')

# we display the modified DataFrame
print(store_items)
```

```python
# We remove the watches and shoes columns
store_items = store_items.drop(['watches', 'shoes'], axis=1)

# we display the modified DataFrame
print(store_items)
```

```python
# We remove the store 2 and store 1 rows
store_items = store_items.drop(['Store 2', 'Store 1'], axis=0)

# we display the modified DataFrame
print(store_items)
```

```python
# We change the column label bikes to hats
store_items = store_items.rename(columns = {'bikes': 'hats'})

# we display the modified DataFrame
print(store_items)
```

```python
# We change the row label from store 3 to last store
store_items = store_items.rename(index = {'store 3': 'last store'})

# we display the modified DataFrame
print(store_items)
```

```python
# We change the row index to be the data in the pants column
store_items = store_items.set_index('pants')

# we display the modified DataFrame
print(store_items)
```
